## 1 · Trace three masks

**By hand: what prints?** (Three masks, one combination, one filter.)

```python
import numpy as np

ages = np.array([34, 61, 47, 70, 29, 55])
treated = np.array([True, True, False, True, False, False])
is_older = (ages >= 50)
older_untreated = is_older & (treated == False)
print(older_untreated.sum())
print(ages[older_untreated])
```

1 and `[55]`: three are 50-plus, but two of those were treated. `treated == False` flips a True/False array, and it is the readable way to say *not* until the course needs more.

## 2 · The mask that was built on the wrong array

A colleague wants the average *weight* of samples whose *volume* exceeds 10, and wrote the chunk below. **It runs without error and prints a plausible number. What did it actually compute?**

```python
import numpy as np

volumes = np.array([8.0, 12.5, 9.5, 15.0, 11.0])
weights = np.array([1.1, 2.3, 1.4, 2.9, 2.0])
is_large = (weights > 10)
if is_large.sum() > 0:
    print(weights[is_large].mean())
else:
    print("no matches")
```

It printed `no matches`: the mask was built on **weights**, none of which exceed 10, instead of on volumes. The code ran without an error and answered a different question, and the `if ... else` guard hid the empty result behind a tidy message. A mask bug of this kind produces no error and often a plausible number, so the code will not tell you. The name is the defense. `is_large` does not say large *what*, and a name such as `has_large_volume` would have made the wrong array visible on the line that built the mask. This course asks you to write the *of what* into the name or the comment.
